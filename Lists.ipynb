```python
orig = 'stressed'
res = ''
for char in orig:
    res = char + res

print(res)
```

```python
data = [1,2,2,5]

cumulative = []
total = 0

for number in data:
    total = total + number
    cumulative.append(total)

print(cumulative)
```

## IF statement

```python
for i in range(1,10):
    if i > 5:
        print(i)
```

```python
if 5 <= 5:
    print("yes")
```

```python
for i in range(1,15):
    if i > 10:
        print(i, "is huge")
    elif i > 5:
        print(i, "is normal")
    elif i == 10:
        print(i, " is 10")
    else:    
        print(i, "is small")
```
