# Functions

```python
def print_date(year, month, day):
    joined = year + '/' + month + '/' + day
    print(joined)
```

```python
print_date('2026','5','20')
```

```python
def print_date_from_numbers(year, month, day):
    joined = str(year) + '/' + str(month) + '/' + str(day)
    print(joined)
```

```python
print_date_from_numbers(2026,5,20)
```

```python
print_date_from_numbers('2026','05','20')
```

```python
def avg(values):
    if len(values) == 0:
        return None
    return sum(values)/len(values)
```

```python
avg([1,2,3,4,5,6])
```

```python
avg([])
```

```python
if avg([]) == None:
    print("surprise!")
```
